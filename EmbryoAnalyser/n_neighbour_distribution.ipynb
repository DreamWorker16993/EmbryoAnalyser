```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import math
```

```python
master_dir = Path(r"C:\Users\ethan\OneDrive - University of Cambridge\fixed_EM\processed")
```

```python
# plot all distributions for one genotype
def plot_geno(genotype: str):
    dir_ = master_dir / genotype
    bins = [0.5*(2*i-1) for i in range(11)]

    row = math.ceil(len(list(dir_.rglob('slow_neighbour_counting.csv'))) / 4)
    fig, ax = plt.subplots(row, 4, figsize = (12, row*3), sharey = True) #sharex = True, 
    
    ax = ax.flatten()
    
    i = 0
    for csv_path in dir_.rglob('slow_neighbour_counting.csv'):
        df = pd.read_csv(csv_path)
        
        # 1. Extract the first 2 characters of the subfolder name (stage)
        folder_prefix = csv_path.parent.name[:2]
        
        # 2. Plot to the current subplot
        sns.histplot(data=df, x="n_neighbours", bins=bins, stat="percent", ax=ax[i])
        
        # 3. Set title on the CURRENT  BEFORE incrementing i
        ax[i].set_title(folder_prefix)
        ax[i].set_xticks(range(0,11,2))
        
        # 4. Increment index for the next iteration
        i += 1
        
    plt.subplots_adjust(hspace=0.6)
    plt.show()
#end def
```

```python
# plot one image for s7 s8 s9 for each genotype
def plot():
    genotypes = ['yw', 'del7', 'sdk5054', 'del15']
    bins = [0.5*(2*i-1) for i in range(11)]
    stages = ['s7','s8','s9']
    fig, ax = plt.subplots(4, 3, figsize = (9, 12), sharey = True) #sharex = True, 
    #ax = ax.flatten()

    for genotype in genotypes:
        row = genotypes.index(genotype)
        dir_ = master_dir / genotype

        #check if an image for each stage already existed
        check = {'s7': False, 's8': False, 's9': False}
        
        for csv_path in dir_.rglob('slow_neighbour_counting.csv'):
            
            # Extract the first 2 characters of the subfolder name (stage)
            prefix = csv_path.parent.name[:2]

            # s10 takes 3 characters. ignore it
            if prefix == 's1':
                continue 
            #end if

            # if this stage is not plotted yet
            if not check[prefix]:
                col = stages.index(prefix)
                df = pd.read_csv(csv_path)
                sns.histplot(data=df, x="n_neighbours", bins=bins, stat="percent", ax=ax[row][col])
                ax[row][col].set_title(genotype + ' ' + prefix)
                ax[row][col].set_xticks([i for i in range(10)])
                check[prefix] = True
            #end if
        #end for
    #end for    
    plt.subplots_adjust(hspace=0.6)
    plt.show()
#end def
```

```python
plot()
```
